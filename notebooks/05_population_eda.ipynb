# Population Dataset (ONS)

Dataset:
Mid-Year Population Estimates (ONS)

Purpose:

```python
from pathlib import Path
import pandas as pd
```

```python
# Project folders

RAW_DATA = Path("../data/raw")
PROCESSED_DATA = Path("../data/processed")

print("Raw folder:", RAW_DATA)
print("Processed folder:", PROCESSED_DATA)
```

```python
# Load MYE4 sheet from ONS population workbook
# This sheet will be audited before deciding whether it becomes
# the population layer for the London Intelligence Dataset.

population_raw = pd.read_excel(
    RAW_DATA / "ons_population_2022.xlsx",
    sheet_name="MYE4",
    header=None
)

population_raw.shape
```

```python
population_raw.head(15)
```

```python
# Load the actual MYE4 table
# Skip metadata rows and use the real table header

population_df = pd.read_excel(
    RAW_DATA / "ons_population_2022.xlsx",
    sheet_name="MYE4",
    header=7
)

population_df.head()
```

```python
print("=" * 60)
print("MYE4 POPULATION DATASET AUDIT")
print("=" * 60)

print(f"Dataset shape: {population_df.shape}")

print("\nColumns:")
print(population_df.columns.tolist())

print("\nData types:")
print(population_df.dtypes)

print("\nMissing values:")
print(population_df.isna().sum())

print("\nGeography types:")
print(population_df["Geography"].value_counts())

print("\nNumber of areas:")
print(population_df["Name"].nunique())

print("\nDuplicate area-year check:")
print(
    population_df.duplicated(
        subset=["Code", "Name"]
    ).sum()
)
```

## Convert Population Dataset from Wide Format to Long Format

The original MYE4 population dataset stores each year as a separate column.
This step reshapes the dataset into a time-series format where each row represents
one borough-area observation for one year.

This structure will make the population layer compatible with the borough-year
property dataset and future data integration steps.

```python
population_long = population_df.melt(
    id_vars=[
        "Code",
        "Name",
        "Geography"
    ],
    var_name="Year",
    value_name="Population"
)

population_long.head()
```

## Reshape Population Data

Convert the ONS population dataset from wide format (years as columns)
into long format (one row per area-year observation).

This structure makes the dataset compatible with time-based analysis
and future merging with the borough-year property dataset.

```python
population_long = population_df.melt(
    id_vars=[
        "Code",
        "Name",
        "Geography"
    ],
    var_name="Year",
    value_name="Population"
)

population_long.head()
```

## Verify Year Column

Check that the Year column is already in numeric format and ready
for time-based analysis.

```python
print(population_long["Year"].dtype)

print(sorted(population_long["Year"].unique()))
```

## Check Geographic Levels

Review the geographic categories available in the ONS population dataset
before filtering London boroughs.

```python
population_long["Geography"].value_counts()
```

## Filter London Boroughs

Select only London Borough observations from the ONS population dataset.

This creates the population layer required for integration with the
borough-year property dataset.

```python
london_population = population_long[
    population_long["Geography"] == "London Borough"
].copy()

london_population.head()
```

```python
london_population.shape
```

```python
london_population["Name"].nunique()
```

```python
sorted(london_population["Year"].unique())
```

```python
print(london_population["Year"].dtype)
```

## Clean Year Column

The Year column contains labels from the original ONS workbook
(e.g. "Mid-2022"). This step extracts the numeric year value
to make the population dataset compatible with other time-based datasets.

```python
london_population["Year"] = (
    london_population["Year"]
    .str.replace("Mid-", "", regex=False)
    .astype(int)
)

london_population.head()
```

## Population Layer Validation

Validate the cleaned London population dataset before integrating it
into the London Intelligence Dataset.

```python
print("=" * 60)
print("LONDON POPULATION LAYER AUDIT")
print("=" * 60)

print(f"Dataset shape: {london_population.shape}")

print(f"\nNumber of boroughs: {london_population['Name'].nunique()}")

print(
    f"Years covered: "
    f"{sorted(london_population['Year'].astype(int).unique().tolist())}"
)

print(
    f"Duplicate Borough-Year pairs: "
    f"{london_population.duplicated(subset=['Name', 'Year']).sum()}"
)

print("\nMissing values:")
print(london_population.isna().sum())
```

```python
OUTPUT_FILE = PROCESSED_DATA / "london_population.csv"

london_population.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Saved to:", OUTPUT_FILE)
```
