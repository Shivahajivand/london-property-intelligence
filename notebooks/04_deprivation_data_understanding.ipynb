### Index of Multiple Deprivation (IMD)

**Objective**

Explore and prepare the IMD dataset for integration into the London Property Intelligence feature store.

```python
import pandas as pd

df_imd = pd.read_excel(
    "../data/raw/ID_2019_for_London.xlsx",
    sheet_name="Borough summary measures",
    header=0
)
```

```python
print("=" * 60)
print("DATASET SHAPE")
print(df_imd.shape)

print("\n" + "=" * 60)
print("COLUMN NAMES")
print(df_imd.columns.tolist())

print("\n" + "=" * 60)
print("DATA TYPES")
df_imd.info()

print("\n" + "=" * 60)
print("FIRST 5 ROWS")
display(df_imd.head())
```

```python
# Remove extra spaces from column names
df_imd.columns = df_imd.columns.str.strip()

print("Columns cleaned successfully.")
```

```python
print("=" * 60)
print("MISSING VALUES")
display(df_imd.isnull().sum())

print("\n" + "=" * 60)
print("UNIQUE VALUES")
display(df_imd.nunique())

print("\n" + "=" * 60)
print("SUMMARY STATISTICS")
display(df_imd.describe())
```

```python
display(

    df_imd
    .sort_values(
        by="IMD - Average score",
        ascending=False
    )[[
        "Local Authority District name (2019)",
        "IMD - Average score"
    ]]
    .head(33)

)
```
