**Objective**

Explore and prepare the IMD dataset for integration into the London Property Intelligence feature store.

```python
import pandas as pd

df_imd = pd.read_excel(
    "../data/raw/ID_2019_for_London.xlsx",
    sheet_name="Borough summary measures",
    header=0
)
```

```python
# Remove extra spaces from column names
df_imd.columns = df_imd.columns.str.strip()

print("Columns cleaned successfully.")
```

```python
print("=" * 60)
print("MISSING VALUES")
display(df_imd.isnull().sum())

print("\n" + "=" * 60)
print("UNIQUE VALUES")
display(df_imd.nunique())

print("\n" + "=" * 60)
print("SUMMARY STATISTICS")
display(df_imd.describe())
```

```python
# Step 1 — Inspect current IMD DataFrame

print("Shape:", df_imd.shape)
print("\nColumns:")
print(df_imd.columns.tolist())

print("\nData types:")
print(df_imd.dtypes)

print("\nFirst 3 rows:")
display(df_imd.head(3))
```

```python
# Step 2 — Validate IMD borough identifiers

print("Unique borough codes:", df_imd["Local Authority District code (2019)"].nunique())
print("Unique borough names:", df_imd["Local Authority District name (2019)"].nunique())

print("\nDuplicate borough codes:")
print(
    df_imd[
        df_imd["Local Authority District code (2019)"].duplicated(keep=False)
    ][
        [
            "Local Authority District code (2019)",
            "Local Authority District name (2019)"
        ]
    ]
)

print("\nDuplicate borough names:")
print(
    df_imd[
        df_imd["Local Authority District name (2019)"].duplicated(keep=False)
    ][
        [
            "Local Authority District code (2019)",
            "Local Authority District name (2019)"
        ]
    ]
)
```

```python
# Step 3 — Create an independent working copy for the IMD feature layer

imd_df = df_imd.copy()

print("Original df_imd shape:", df_imd.shape)
print("New imd_df shape:", imd_df.shape)
print("\nDataFrames are independent copies.")
```

```python
# Step 4 — Review IMD feature columns before standardisation

for i, column in enumerate(imd_df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Step 5 — Standardise IMD column names

imd_df = imd_df.rename(columns={
    "Local Authority District code (2019)": "Borough_Code",
    "Local Authority District name (2019)": "Borough_Name",
    "IMD - Average rank": "IMD_Average_Rank",
    "IMD - Rank of average rank": "IMD_Rank_of_Average_Rank",
    "IMD - Average score": "IMD_Average_Score",
    "IMD - Rank of average score": "IMD_Rank_of_Average_Score",
    "IMD - Proportion of LSOAs in most deprived 10% nationally": "IMD_Proportion_Most_Deprived_10pct",
    "IMD - Rank of proportion of LSOAs in most deprived 10% nationally": "IMD_Rank_of_Proportion_Most_Deprived_10pct",
    "IMD 2019 - Extent": "IMD_Extent",
    "IMD 2019 - Rank of extent": "IMD_Rank_of_Extent",
    "IMD 2019 - Local concentration": "IMD_Local_Concentration",
    "IMD 2019 - Rank of local concentration": "IMD_Rank_of_Local_Concentration"
})

print("Columns after standardisation:")
print(imd_df.columns.tolist())
```

```python
# Step 6 — Add the reference year for the IMD dataset

imd_df["Year"] = 2019

print("Shape:", imd_df.shape)
print("\nYear values:")
print(imd_df["Year"].unique())
```

```python
# Step 7 — Final structural check before export

print("Shape:", imd_df.shape)

print("\nColumns:")
for i, column in enumerate(imd_df.columns, start=1):
    print(f"{i}. {column}")

print("\nData types:")
print(imd_df.dtypes)

print("\nFirst 3 rows:")
display(imd_df.head(3))
```

```python
# Step 8 — Final pre-export audit

print("=== IMD PRE-EXPORT AUDIT ===")

print("Shape:", imd_df.shape)
print("Unique borough codes:", imd_df["Borough_Code"].nunique())
print("Unique borough names:", imd_df["Borough_Name"].nunique())
print("Unique years:", imd_df["Year"].unique())
print("Duplicate full rows:", imd_df.duplicated().sum())
print("Number of columns:", imd_df.shape[1])
```

```python
# Step 9 — Export the independent IMD feature layer

output_path = "../data/processed/london_imd.csv"

imd_df.to_csv(output_path, index=False)

print(f"IMD feature layer exported successfully to:")
print(output_path)
```

```python
# Step 10 — Round-trip validation of exported IMD layer

imd_check = pd.read_csv(output_path)

print("=== EXPORTED FILE VALIDATION ===")

print("Shape:", imd_check.shape)

print("\nColumns:")
print(imd_check.columns.tolist())

print("\nMissing values:")
print("Total:", imd_check.isna().sum().sum())

print("\nUnique borough codes:", imd_check["Borough_Code"].nunique())
print("Unique borough names:", imd_check["Borough_Name"].nunique())
print("Unique years:", imd_check["Year"].unique())

print("\nDuplicate full rows:", imd_check.duplicated().sum())

print("\nFirst 3 rows:")
display(imd_check.head(3))
```
