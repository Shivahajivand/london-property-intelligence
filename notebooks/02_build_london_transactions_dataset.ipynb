# Building the Multi-Year London Dataset

## Objective

Build a clean multi-year London property transaction dataset (2018–2025)
from the HM Land Registry Price Paid Data.

This dataset will serve as the foundation for all subsequent feature
engineering, borough-level aggregation, and machine learning models.

## Workflow

HM Land Registry Price Paid Data (2018–2025)

↓

Merge yearly datasets

↓

Filter Greater London transactions

↓

Save processed dataset

## Output

The processed dataset will be saved as:

`data/processed/london_transactions_2018_2025.csv`

```python
import pandas as pd
from pathlib import Path
```

```python
# Project folders

RAW_DATA = Path("../data/raw")
PROCESSED_DATA = Path("../data/processed")

print("Raw folder:", RAW_DATA)
print("Processed folder:", PROCESSED_DATA)
```

```python
# List all CSV files

csv_files = sorted(RAW_DATA.glob("*.csv"))

print(f"Number of files found: {len(csv_files)}")

for file in csv_files:
    print(file.name)
```

```python
# Official HM Land Registry column names

column_names = [
    "Transaction unique identifier",
    "Price",
    "Date of Transfer",
    "Postcode",
    "Property Type",
    "Old/New",
    "Duration",
    "PAON",
    "SAON",
    "Street",
    "Locality",
    "Town/City",
    "District",
    "County",
    "PPD Category Type",
    "Record Status"
]
```

```python
# Read one sample file

sample_file = csv_files[0]

df = pd.read_csv(
    sample_file,
    header=None,
    names=column_names
)

df.head()
```

```python
print(sample_file.name)
```

```python
# Read all yearly files

dataframes = []

for file in csv_files:

    # Read file
    df = pd.read_csv(
        file,
        header=None,
        names=column_names
    )

    # Extract year from filename
    year = int(file.stem.split("-")[1])

    # Add Year column
    df["Year"] = year

    # Save DataFrame
    dataframes.append(df)

    print(f"{file.name} -> {len(df):,} rows")
```

```python
print(len(dataframes))
```

```python
transactions_all_years = pd.concat(
    dataframes,
    ignore_index=True
)
```

```python
print(transactions_all_years.shape)
```

```python
transactions_all_years.head()
```

```python
transactions_all_years["Year"].value_counts().sort_index()
```

```python
# Filter Greater London transactions

london_transactions = transactions_all_years[
    transactions_all_years["County"] == "GREATER LONDON"
].copy()
```

```python
print(f"Number of London transactions: {len(london_transactions):,}")
```

```python
london_transactions.head()
```

```python
# Save processed London dataset

output_file = PROCESSED_DATA / "london_transactions_2018_2025.csv"

london_transactions.to_csv(
    output_file,
    index=False
)

print(f"Saved to:\n{output_file}")
```
